```python

import json
import os

from src.inner import tide_gauge_station

# read the tide gauge station data, make csv file with id, lat, lon

step_length = 10
start_year = 2012
end_year = 2022
out_dir = "../../output/PR_bonn"
# The tide gauge stations can be found here
station_list_file_path = "../../data/rlr_monthly/filelist.txt"
land_path = "../../data/ne_10m_land/ne_10m_land.shp"  # used as background for plotting
ocean_path = "../../data/ocean_polygon/ne_10m_ocean.shp"

# ----------------------------------------------------
# Setup
if not os.path.exists(out_dir):
    os.makedirs(out_dir)
station_info_path = os.path.join(out_dir, "station_info.txt")
stations = tide_gauge_station.read_and_create_stations(station_list_file_path, station_info_path)
stations = tide_gauge_station.filter_stations_for_time_step(stations, start_year, end_year)
# write csv with the tide gauge stations that exist in the current timestep

# read station solution and check that all stations in the solution are in the station list and not more
solution_path = "../../output/results/results/Voronoi_based_equitable_connectedclustering/2012_2022/solution250.json"
with open(solution_path, "r") as solution_file:
    solution = json.load(solution_file)

# check that all stations in the solution are in the station list and not more
stations_in_solution = set()
stations_to_remove = []
for center_id, station_id_list in solution.items():
    if int(center_id) not in stations.keys():
        print(f"center {center_id} not in station list")
    stations_in_solution.add(int(center_id))
    for station_id in station_id_list:
        if int(station_id) not in stations.keys():
            print(f"station {station_id} not in station list")
        stations_in_solution.add(int(station_id))
for station_id in stations.keys():
    if station_id not in stations_in_solution:
        print(f"station {station_id} not in solution list")
        stations_to_remove.append(station_id)
# remove stations that are not in the solution
for station_id in stations_to_remove:
    del stations[station_id]

print(len(stations))
print(len(stations_in_solution))
print(len(solution))

with open(os.path.join(out_dir, "station_list.csv"), "w") as output_file:
    for station in stations.values():
        output_file.write(f"{station.id};{station.latitude};{station.longitude}\n")

```
